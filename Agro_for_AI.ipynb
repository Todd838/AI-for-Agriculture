```python
from pathlib import Path
import warnings
import numpy as np
import pandas as pd
import cv2
import rasterio
warnings.filterwarnings("ignore", message=".*geotransform.*")
from skimage.feature import graycomatrix, graycoprops
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score
from sklearn.preprocessing import LabelEncoder
from sklearn.decomposition import PCA
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier, HistGradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

RANDOM_STATE = 42
N_SPLITS = 5
GLCM_DISTANCES = [1, 2]
GLCM_ANGLES = [0, np.pi / 4, np.pi / 2]
GLCM_LEVELS = 32
HS_DROP_FIRST = 10
HS_DROP_LAST = 14
HS_REP_BANDS = 10
PCA_N_COMPONENTS = 15

np.random.seed(RANDOM_STATE)

# Paths: ROOT_DIR, DATA_DIR, VAL_DIR, USE_FLAT_STRUCTURE, COMBINE_TRAIN_VAL, EVALUATE_ON_VAL, OUTPUT_DIR, FEATURES_CSV_PATH, BEST_MODEL_PATH - must be set externally

def parse_label_from_filename(filename: str) -> str:
    stem = Path(filename).stem
    if "_" not in stem:
        raise ValueError(f"Filename does not match '<class>_<fileid>': {filename}")
    return stem.split("_", 1)[0]

def find_first_file(folder: Path, exts=(".png", ".jpg", ".jpeg", ".tif", ".tiff")) -> Path | None:
    if not folder.exists():
        return None
    for p in sorted(folder.iterdir()):
        if p.is_file() and p.suffix.lower() in exts:
            return p
    return None

def build_index_dataframe(data_dir: Path) -> pd.DataFrame:
    rows = []
    for sample_dir in sorted([p for p in data_dir.iterdir() if p.is_dir()]):
        rgb_path = find_first_file(sample_dir / "RGB", exts=(".png", ".jpg", ".jpeg"))
        ms_path = find_first_file(sample_dir / "MS", exts=(".tif", ".tiff"))
        hs_path = find_first_file(sample_dir / "HS", exts=(".tif", ".tiff"))
        if rgb_path is None and ms_path is None and hs_path is None:
            continue
        label_source = rgb_path or ms_path or hs_path
        label = parse_label_from_filename(label_source.name)
        rows.append({"sample_id": sample_dir.name, "rgb_path": str(rgb_path) if rgb_path else None,
                    "ms_path": str(ms_path) if ms_path else None, "hs_path": str(hs_path) if hs_path else None, "label": label})
    df = pd.DataFrame(rows)
    if df.empty:
        raise RuntimeError(f"No samples found under: {data_dir.resolve()}")
    return df

def build_index_dataframe_flat(data_dir: Path) -> pd.DataFrame:
    def collect_files(folder: Path, exts: tuple) -> dict:
        out = {}
        if folder.exists():
            for p in folder.iterdir():
                if p.is_file() and p.suffix.lower() in exts:
                    out[p.stem] = p
        return out
    rgb_files = collect_files(data_dir / "RGB", (".png", ".jpg", ".jpeg"))
    ms_files = collect_files(data_dir / "MS", (".tif", ".tiff"))
    hs_files = collect_files(data_dir / "HS", (".tif", ".tiff"))
    all_stems = set(rgb_files) | set(ms_files) | set(hs_files)
    if not all_stems:
        raise RuntimeError(f"No image files found under: {data_dir.resolve()}")
    rows = []
    for stem in sorted(all_stems):
        rgb_path, ms_path, hs_path = rgb_files.get(stem), ms_files.get(stem), hs_files.get(stem)
        label_source = rgb_path or ms_path or hs_path
        label = parse_label_from_filename(label_source.name)
        rows.append({"sample_id": stem, "rgb_path": str(rgb_path) if rgb_path else None,
                    "ms_path": str(ms_path) if ms_path else None, "hs_path": str(hs_path) if hs_path else None, "label": label})
    return pd.DataFrame(rows)

def _build_index(dir_path: Path) -> pd.DataFrame:
    return build_index_dataframe_flat(dir_path) if USE_FLAT_STRUCTURE else build_index_dataframe(dir_path)

df_train = _build_index(DATA_DIR) if DATA_DIR.exists() else pd.DataFrame()
if COMBINE_TRAIN_VAL and VAL_DIR.exists():
    df_val = _build_index(VAL_DIR)
    df_val["sample_id"] = "val_" + df_val["sample_id"].astype(str)
    df_index = pd.concat([df_train.assign(sample_id="train_" + df_train["sample_id"].astype(str)), df_val], ignore_index=True) if not df_train.empty else df_val
    print("STEP 2 — Combined train + val for training")
else:
    df_index = df_train
if df_index.empty:
    raise RuntimeError(f"No samples found. Check DATA_DIR and VAL_DIR")
print("STEP 2 — Dataset index:"); print(df_index.head()); print(df_index[["rgb_path", "ms_path", "hs_path"]].isna().sum()); print(df_index["label"].value_counts())

def safe_skew(x: np.ndarray, eps: float = 1e-8) -> float:
    x = x.astype(np.float64); mu, sigma = np.mean(x), np.std(x) + eps
    return float(np.mean(((x - mu) / sigma) ** 3))

def nan_safe_stats(x: np.ndarray) -> tuple:
    x = x.astype(np.float64)[np.isfinite(x)]
    return (float(np.mean(x)), float(np.std(x)), safe_skew(x)) if x.size else (0.0, 0.0, 0.0)

def extract_rgb_features(rgb_path) -> dict:
    feats = {}
    if rgb_path is None:
        for ch in ["b","g","r"]: feats.update({f"rgb_{ch}_mean":0,f"rgb_{ch}_std":0,f"rgb_{ch}_skew":0})
        for p in ["contrast","homogeneity","energy","correlation"]: feats[f"glcm_{p}"]=0
        return feats
    img = cv2.imread(rgb_path, cv2.IMREAD_COLOR)
    if img is None: raise ValueError(f"Could not read: {rgb_path}")
    for n,ch in zip(["b","g","r"],[img[:,:,0].ravel(),img[:,:,1].ravel(),img[:,:,2].ravel()]):
        m,s,k = nan_safe_stats(ch); feats[f"rgb_{n}_mean"],feats[f"rgb_{n}_std"],feats[f"rgb_{n}_skew"]=m,s,k
    gray = (cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float32)/255*(GLCM_LEVELS-1)).astype(np.uint8)
    glcm = graycomatrix(gray, distances=GLCM_DISTANCES, angles=GLCM_ANGLES, levels=GLCM_LEVELS, symmetric=True, normed=True)
    for p in ["contrast","homogeneity","energy","correlation"]: feats[f"glcm_{p}"]=float(np.mean(graycoprops(glcm,p)))
    return feats

def extract_ms_features(ms_path) -> dict:
    feats = {}
    if ms_path is None:
        for i in range(1,6): feats[f"ms_band{i}_mean"],feats[f"ms_band{i}_std"]=0,0
        for idx in ["ndvi","ndre","gndvi","savi","evi"]: feats[f"ms_{idx}_mean"],feats[f"ms_{idx}_std"],feats[f"ms_{idx}_skew"]=0,0,0
        return feats
    with rasterio.open(ms_path) as src: ms = src.read().astype(np.float32)
    if ms.shape[0]<5: raise ValueError(f"Expected ≥5 MS bands: {ms_path}")
    B,G,R,RE,NIR = ms[:5]; eps=1e-8
    for i in range(5): m,s,_=nan_safe_stats(ms[i].ravel()); feats[f"ms_band{i+1}_mean"],feats[f"ms_band{i+1}_std"]=m,s
    for nm,arr in [("ndvi",(NIR-R)/(NIR+R+eps)),("ndre",(NIR-RE)/(NIR+RE+eps)),("gndvi",(NIR-G)/(NIR+G+eps)),
                   ("savi",1.5*(NIR-R)/(NIR+R+0.5+eps)),("evi",2.5*(NIR-R)/(NIR+6*R-7.5*B+1+eps))]:
        m,s,k=nan_safe_stats(arr.ravel()); feats[f"ms_{nm}_mean"],feats[f"ms_{nm}_std"],feats[f"ms_{nm}_skew"]=m,s,k
    return feats

def extract_hs_features(hs_path) -> tuple:
    feats = {}
    if hs_path is None:
        for k in ["hs_curve_mean","hs_curve_std","hs_curve_skew","hs_d1_mean","hs_d2_mean","hs_auc"]: feats[k]=0
        for i in range(HS_REP_BANDS): feats[f"hs_repband{i+1}_mean"],feats[f"hs_repband{i+1}_std"]=0,0
        return feats, np.array([], dtype=np.float32)
    with rasterio.open(hs_path) as src: hs = src.read().astype(np.float32)
    n=hs.shape[0]
    if n<=HS_DROP_FIRST+HS_DROP_LAST+1: raise ValueError(f"Too few HS bands: {hs_path}")
    hs_trim=hs[HS_DROP_FIRST:n-HS_DROP_LAST]; curve=np.mean(hs_trim.reshape(hs_trim.shape[0],-1),axis=1)
    feats["hs_curve_mean"],feats["hs_curve_std"],feats["hs_curve_skew"]=float(np.mean(curve)),float(np.std(curve)),safe_skew(curve)
    d1,d2=np.diff(curve),np.diff(curve,n=2)
    feats["hs_d1_mean"],feats["hs_d2_mean"]=(float(np.mean(d1)) if d1.size else 0),(float(np.mean(d2)) if d2.size else 0)
    feats["hs_auc"]=float(np.trapezoid(curve))
    for i,b in enumerate(np.linspace(0,hs_trim.shape[0]-1,HS_REP_BANDS).astype(int),1):
        m,s,_=nan_safe_stats(hs_trim[b].ravel()); feats[f"hs_repband{i}_mean"],feats[f"hs_repband{i}_std"]=m,s
    return feats, curve.astype(np.float32)

def build_features_dataframe(df_index):
    rows=[]
    for _,r in df_index.iterrows():
        row={"sample_id":r["sample_id"],"label":r["label"]}
        row.update(extract_rgb_features(r["rgb_path"])); row.update(extract_ms_features(r["ms_path"]))
        hf,hc=extract_hs_features(r["hs_path"]); row.update(hf)
        for i,v in enumerate(hc): row[f"hs_curve_{i:03d}"]=float(v)
        rows.append(row)
    df=pd.DataFrame(rows)
    cc=[c for c in df.columns if c.startswith("hs_curve_")]
    if cc: df[cc]=df[cc].fillna(0); df=df.fillna(0)
    return df

df_features=build_features_dataframe(df_index)
print("\nSTEP 4 — Feature matrix:", df_features.shape); df_features.to_csv(FEATURES_CSV_PATH, index=False)

df=pd.read_csv(FEATURES_CSV_PATH); y=df["label"].copy(); X=df.drop(columns=["label","sample_id"],errors="ignore")
curve_cols=[c for c in X.columns if c.startswith("hs_curve_")]; other_cols=[c for c in X.columns if c not in curve_cols]
pca_preprocessor=ColumnTransformer([("hs_pca",PCA(n_components=PCA_N_COMPONENTS,random_state=RANDOM_STATE),curve_cols),("other","passthrough",other_cols)],remainder="drop",verbose_feature_names_out=False) if curve_cols and len(curve_cols)>=PCA_N_COMPONENTS else ColumnTransformer([("all","passthrough",list(X.columns))],remainder="drop",verbose_feature_names_out=False)

le=LabelEncoder(); y_enc=le.fit_transform(y); n_classes=len(le.classes_); is_binary=n_classes==2
skf=StratifiedKFold(n_splits=N_SPLITS,shuffle=True,random_state=RANDOM_STATE)
models={"HistGradientBoosting":HistGradientBoostingClassifier(random_state=RANDOM_STATE),"RandomForest":RandomForestClassifier(n_estimators=300,random_state=RANDOM_STATE,n_jobs=-1,class_weight="balanced"),"ExtraTrees":ExtraTreesClassifier(n_estimators=500,random_state=RANDOM_STATE,n_jobs=-1,class_weight="balanced"),"DecisionTree":DecisionTreeClassifier(random_state=RANDOM_STATE,class_weight="balanced")}

def evaluate_model_cv(model,X,y_enc):
    accs,f1s,aucs=[],[],[]
    for ti,vi in skf.split(X,y_enc):
        pipe=Pipeline([("prep",pca_preprocessor),("clf",model)])
        pipe.fit(X.iloc[ti],y_enc[ti])
        accs.append(accuracy_score(y_enc[vi],pipe.predict(X.iloc[vi])))
        f1s.append(f1_score(y_enc[vi],pipe.predict(X.iloc[vi]),average="macro"))
        if is_binary and hasattr(pipe,"predict_proba"): aucs.append(roc_auc_score(y_enc[vi],pipe.predict_proba(X.iloc[vi])[:,1]))
    out={"acc_mean":np.mean(accs),"acc_std":np.std(accs),"f1_mean":np.mean(f1s),"f1_std":np.std(f1s)}
    if is_binary: out["auc_mean"],out["auc_std"]=np.nanmean(aucs),np.nanstd(aucs)
    return out

results=pd.DataFrame([{"model":n,**evaluate_model_cv(m,X,y_enc)} for n,m in models.items()]).sort_values("f1_mean",ascending=False)
print("\nCV Results:"); print(results)

plot_df=results.melt(id_vars=["model"],value_vars=["acc_mean","f1_mean"],var_name="metric",value_name="score")
fig,ax=plt.subplots(figsize=(10,5)); sns.barplot(data=plot_df,x="model",y="score",hue="metric")
ax.set_title("Model Comparison (Mean CV Scores)"); plt.xticks(rotation=20,ha="right"); plt.tight_layout()
plt.savefig(OUTPUT_DIR/"model_comparison.png",dpi=150); plt.show()

best_name=results.iloc[0]["model"]; print(f"\nBest model: {best_name}")
pipe=Pipeline([("prep",pca_preprocessor),("clf",models[best_name])]); pipe.fit(X,y_enc)
joblib.dump({"pipeline":pipe,"label_encoder":le,"curve_cols":curve_cols,"other_cols":other_cols,"classes":list(le.classes_)},BEST_MODEL_PATH)
print(f"Saved to {BEST_MODEL_PATH}")
```
